```python
import pandas as pd

sales = pd.read_csv("sales 1.csv")

products = pd.read_csv("products.csv")

customers = pd.read_json("customers.json")

print("Sales Shape:", sales.shape)
print("Products Shape:", products.shape)
print("Customers Shape:", customers.shape)
```

```python
customers.head()
```

```python
print(
    "Sales Duplicates:",
    sales.duplicated().sum()
)

print(
    "Customer Duplicates:",
    customers.duplicated().sum()
)
```

```python
sales["Timestamp"] = pd.to_datetime(
    sales["Timestamp"]
)
```

```python
sales["Revenue"] = (
    sales["SalesAmount"]
    *
    sales["Quantity"]
)
```

```python
sales_product = sales.merge(
    products,
    on="ProductID"
)
```

```python
retail = sales_product.merge(
    customers,
    on="CustomerID"
)
```

```python
retail.head()
```

```python
total_revenue = retail[
    "Revenue"
].sum()

print(total_revenue)
```

```python
retail[
"CustomerID"
].nunique()
```

```python
retail.groupby(
"ProductName"
)["Revenue"].sum().sort_values(
ascending=False
).head(10)
```

```python
retail.groupby(
"Region"
)["Revenue"].sum()
```

```python
clv = retail.groupby(
"CustomerID"
)["Revenue"].sum()
```

```python
sales["SalesAmount"] = sales["SalesAmount"].fillna(
    sales["SalesAmount"].median()
)
```

```python
sales["Quantity"] = sales["Quantity"].fillna(
    sales["Quantity"].median()
)
```

```python
customers = customers.drop_duplicates()
```

```python
customers["LastName"] = customers[
    "LastName"
].fillna("Unknown")
```

```python
customers["Gender"] = customers[
    "Gender"
].replace({
        "M":"Male",
        "male":"Male",
        "F":"Female",
        "female":"Female"
})
```

```python
customers["Region"] = customers[
    "Region"
].replace({
    "Ohho":"Ohio",
    "Nw York":"New York",
    "New Yorkk":"New York",
    "Californiya":"California",
    "Texaz":"Texas",
    "california":"California",
    "NY":"New York"
})
```

```python
sales["Timestamp"] = pd.to_datetime(
    sales["Timestamp"]
)
```

```python
sales["Revenue"] = (
    sales["SalesAmount"]
    *
    sales["Quantity"]
)
```

```python
sales_product = sales.merge(
    products,
    on="ProductID"
)
```

```python
retail = sales_product.merge(
    customers,
    on="CustomerID"
)
```

```python
retail.head()
```

```python
total_revenue = retail[
    "Revenue"
].sum()

print(total_revenue)
```

```python
retail[
"CustomerID"
].nunique()
```

```python
retail.groupby(
"ProductName"
)["Revenue"].sum().sort_values(
ascending=False
).head(10)
```

```python
retail.groupby(
"Region"
)["Revenue"].sum()
```

```python
clv = retail.groupby(
"CustomerID"
)["Revenue"].sum()
```
